```python
order_items['returned'] = order_items['returned'].astype("string").str.strip().str.title().replace(
    {"N":0,
     "No":0,
     "0":0,
     pd.NA:0,
     "False":0,
     "Y":1,
     "Yes":1,
     "True":1,
     "1":1,
     
     })
```

```python
order_items['quantity'] = pd.to_numeric(order_items['quantity'], errors="coerce").astype("Int64")
order_items['discount_raw'] =order_items['discount']
text = order_items['discount'].astype("string").str.strip()
d = pd.to_numeric(text.replace("%","",regex=False),errors="coerce")
d= d.where(~d.between(5,30), d/100)
is_valid = d.between(0,0.30)
order_items['discount'] = d.where(is_valid)
order_items['gross_sales'] = order_items['quantity'] * order_items['unit_price']
order_items['net_sales'] = order_items['gross_sales'] * (1-order_items['discount'])
profit = order_items['net_sales'] - order_items['cost']
raw_est = 1 - (order_items["profit"] + order_items["cost"]) / (order_items['quantity'] * order_items["unit_price"])
snapped = ((raw_est / 0.05).round() * 0.05).round(2)
close = (raw_est - snapped).abs() < 0.02 

# order_items['discount'] = order_items['discount'].where(
#     valid_discount,
#     pd.NA
# )
# order_items['quantity']
order_items= order_items[(order_items['quantity'] > 0)]
quantity = (order_items['cost'] / order_items['unit_price'])

order_items['quantity'] = order_items['quantity'].fillna(quantity)

unit_cost = (order_items['cost'] + order_items['profit'] / order_items['quantity'] * (1-order_items['discount']) )
order_items['unit_price'] = order_items['unit_price'].fillna(unit_cost)
order_items['profit']  = order_items['profit'].fillna(profit)
fill = order_items["discount"].isna() & snapped.between(0, 0.30) & close
order_items["discount_recovered"] = fill
order_items.loc[fill, "discount"] = snapped[fill]
```
